**[Task 2.1.]()** Implement Otsu thresholding as a re-usable function:

```python
def otsu_thresholding(imgf):
    img8 = (imgf * 255).round().astype(np.uint8)
    scores = np.full(256, np.inf)

    for T in range(1, 256):
        class1 = img8[img8 < T]
        class2 = img8[img8 >= T]

        if class1.size > 0 and class2.size > 0:
            scores[T] = (
                class1.size * class1.var()
                + class2.size * class2.var()
            )

    T_star = np.argmin(scores)
    seg = img8 >= T_star
    return seg
```

**[Task 2.2.]()** Implement a re-usable function to compute the Dice coefficient:

i just used bins instead of official images 

```python
 def compute_dice(bin1, bin2):
    assert bin1.dtype == bool
    assert bin2.dtype == bool

    dice = (2 * (bin1 * bin2).sum()) / (bin1.sum() + bin2.sum())

    return dice

```

**[Task 2.3.]()** Test your implementations from above:

```python
imgf = plt.imread('data/NIH3T3/im/dna-0.png')
ground_image = plt.imread('data/NIH3T3/gt/0.png') > 0

seg = otsu_thresholding(imgf)
dice = compute_dice(seg, ground_image)

print("Dice coefficient:", dice)

plt.imshow(seg, cmap='gray')
plt.show()
```

**[Task 2.4.]()** Perform batch processing:

```python
dice_scores = []

for i in [28, 29, 33, 44, 46, 49]:
    imgf = plt.imread(f'data/NIH3T3/im/dna-{i}.png')
    ground_image = plt.imread(f'data/NIH3T3/gt/{i}.png') > 0

    seg = otsu_thresholding(imgf)
    dice = compute_dice(seg, ground_image)

    dice_scores.append(dice)
    print(f'Image {i}: {dice}')

print('Mean Dice coefficient:', np.mean(dice_scores))
```
