**[Task 2.2.]()** Implement the two functions `sobel_h` and `sobel_v` for horizontal and vertical $3 \times 3$ Sobel filtering, respectively:

```python
def sobel_h(img):
    kernel = np.array([
        [-1, 0, 1],
        [-2, 0, 2],
        [-1, 0, 1]
    ])
    kernel = np.flip(kernel)

    result = np.zeros(img.shape)

    for row in range(1, img.shape[0] - 1):
        for col in range(1, img.shape[1] - 1):
            patch = img[row-1:row+2, col-1:col+2]
            result[row, col] = np.sum(patch * kernel)

    return result


def sobel_v(img):
    kernel = np.array([
        [-1, -2, -1],
        [ 0,  0,  0],
        [ 1,  2,  1]
    ])
    kernel = np.flip(kernel)

    result = np.zeros(img.shape)

    for row in range(1, img.shape[0] - 1):
        for col in range(1, img.shape[1] - 1):
            patch = img[row-1:row+2, col-1:col+2]
            result[row, col] = np.sum(patch * kernel)

    return result
```

**[Task 2.3.]()** Test your implementations from above:

```python
gx = sobel_h(img)
gy = sobel_v(img)

plt.figure()
plt.imshow(gx, cmap='gray')
plt.colorbar()
plt.title('Horizontal derivative')
plt.show()

plt.figure()
plt.imshow(gy, cmap='gray')
plt.colorbar()
plt.title('Vertical derivative')
plt.show()
```

**[Task 2.4.]()** Compute the image gradient magnitude:

```python
magnitude = np.sqrt(gx**2 + gy**2)

plt.figure()
plt.imshow(magnitude, cmap='gray')
plt.colorbar()
plt.title('Gradient magnitude')
plt.show()
```
