```python
import numpy as np
import matplotlib.pyplot as plt
```

**[Task 1.2 (a).]()** Use `plt.imread` to load an image:

```python
image = plt.imread('data/NIH3T3/im/dna-0.png')
```

**[Task 1.2 (b).]()** Show the image:

```python

plt.figure()
plt.imshow(image)
plt.colorbar()
```

**[Task 1.2 (d).]()** Binarize the image by applying an intensity threshold:

```python
threshold = 0.10
bin_image = image >= threshold 

plt.figure(figsize=(15, 8))
plt.imshow(bin_image, cmap='gray', vmin=0, vmax=1)
plt.show()
```

**[Task 1.2 (e).]()** Load the ground truth:

```python
ground_image = plt.imread('data/NIH3T3/gt/0.png')
ground_image.dtype
bin_image.dtype
```

Compute and print the Dice coefficient:

```python
Dice = (2 * (ground_image * bin_image).sum() ) / (ground_image.sum() + bin_image.sum())
print(Dice)
```
